#Daily Challenge: Getting Started with LangChain Conversational Memory


```python
# 1. Importing the Memory Module

# Importing the ConversationBufferMemory class from the memory module
from langchain.memory import ConversationBufferMemory

# This imports the ConversationBufferMemory class, which we'll use to store and retrieve conversation history.
```

```python
# 2. Initializing the Memory

# Creating an instance of the memory buffer
memory = ConversationBufferMemory()

# This is setting up the memory so it can start storing conversation context (what the user and bot say).
```

```python
# 3. Simulating a First Interaction

# Defining the first user input
user_input = "Hello, how are you?"

# Defining the bot's response
bot_output = "I'm fine, thank you. How can I assist you today?"

# Saving the first exchange into memory
memory.save_context(
    {"input": user_input},
    {"output": bot_output}
)

# This is storing the first exchange into memory so it can be remembered later.
```

```python
# 4. Simulating a Follow-up Message

# Defining the follow-up user input
user_input = "Tell me a joke."

# Defining the bot's response
bot_output = "Why did the chicken cross the road? To get to the other side."

# Saving the second exchange into memory
memory.save_context(
    {"input": user_input},
    {"output": bot_output}
)

# This is adding the second user-bot exchange into the same memory buffer.
```

```python
# 5. Retrieving the Conversation History

# Loading the stored memory into a variable
conversation_history = memory.load_memory_variables({})

# This is retrieving everything stored in memory so far (all the exchanges) and saving it into the conversation_history variable.
```

```python
# 6. Printing the Memory

# Printing the full conversation history
print(conversation_history)

# This is displaying how LangChain stored the conversation — you should see both exchanges in the output.
```
